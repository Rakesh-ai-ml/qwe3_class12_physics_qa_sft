```python
import os
import json
from datasets import load_dataset, Dataset
from src.NcrtPhysicsQwen3.logger import logger
from src.NcrtPhysicsQwen3.config.configuration import ModelEvaluationConfig

from ragas import evaluate
from ragas.metrics import FactualCorrectness
# from ragas.metrics.collections import  FactualCorrectness
# from ragas.llms import LangchainLLMWrapper
# from langchain_openai import ChatOpenAI

import wandb
from dotenv import load_dotenv
load_dotenv()

from ragas.llms import llm_factory
from openai import AsyncOpenAI 




class DatasetPrep:
    def __init__(self):
        self.dataset_path = "Rakesh7n/ncrt_eval_outputs"

    def combine_cols(self, example):
        ans = example['Answer']
        exp = example['Explanation']
        return {"reference": f"Answer: {ans}\nExplanation: {exp}"}

    def load_test_data(self):
        ds = load_dataset(str(self.dataset_path), split="train")
        ds = ds.map(self.combine_cols)
        ds = ds.remove_columns(['Explanation', 'Question', 'Answer', 'messages'])
        ds=ds.select(range(1))
        eval_ds =Dataset.from_dict({
                "user_input": ds["formatted_prompt"],   # maps to user_input
                "response": ds["adapter_output"],        # maps to response
                "reference": ds["reference"],            # maps to reference
            })
        return ds

        


class ModelEvaluation:

    def __init__(self, config: ModelEvaluationConfig):
        self.config = config
        nim_client = AsyncOpenAI (
                base_url="https://integrate.api.nvidia.com/v1",
                api_key=os.getenv("NVIDIA_API_KEY"),
            )
        self.llm = llm_factory("meta/muse-glimmer-30b", client=nim_client)   
        self.ds = DatasetPrep().load_test_data()

    def run_evaluation(self, output_col="adapter_output"):

        metrics = [FactualCorrectness(llm=self.llm)]
        result = evaluate(
            dataset=self.ds,
            metrics=metrics,
            llm=self.llm,
            column_map={
                "user_input": "formatted_prompt",
                "response": "adapter_output",
                "reference": "reference",
            },
        )

        df = result.to_pandas()

        # --- Save JSON ---
        json_results = df.to_dict(orient="records")
        output_path = os.path.join(self.config.root_dir, "ragas_results.json")
        with open(output_path, "w", encoding="utf-8") as f:
            json.dump(json_results, f, indent=2, ensure_ascii=False)
        logger.info(f"Results saved to {output_path}")

        # --- Summary ---
        col = "factual_correctness"
        avg = float(df[col].mean())
        pass_rate = float((df[col] >= 0.5).mean())
        logger.info(f"Factual Correctness: avg={avg:.3f}, pass_rate={pass_rate:.1%}")

        # --- Wandb ---
        wandb.init(project="ncrt-physics-eval", name=f"ragas-{output_col}")
        wandb.log({"factual_correctness_avg": avg, "factual_correctness_pass_rate": pass_rate})

        table = wandb.Table(
            data=df[["question", "answer", "ground_truth", col]].values.tolist(),
            columns=["question", "answer", "ground_truth", "factual_correctness"],
        )
        wandb.log({"sample_table": table})
        wandb.finish()   

# --- Run ---
# config = ModelEvaluationConfig(root_dir="./")
# evaluator = ModelEvaluation(config)
# evaluator.run_evaluation()   

# import os
# import json
# from datasets import load_dataset
# from deepeval import evaluate
# from deepeval.test_case import LLMTestCase,  LLMTestCaseParams
# from deepeval.metrics import AnswerRelevancyMetric, GEval
# from deepeval.models import LocalModel
# from src.NcrtPhysicsQwen3.logger import logger
# from src.NcrtPhysicsQwen3.config.configuration import ModelEvaluationConfig
# import wandb
# from dotenv import load_dotenv
# load_dotenv()

# # class NIMModel(LocalModel):
# #     def __init__(self, model_name="meta/muse-glimmer-30b", temperature=0):
# #         self.model_name = model_name
# #         self.temperature = temperature
# #         from openai import OpenAI, AsyncOpenAI
# #         self.model = OpenAI(
# #             base_url=os.environ["LOCAL_MODEL_BASE_URL"],
# #             api_key=os.environ["NVIDIA_API_KEY"],
# #             timeout=120,
# #             max_retries=5,
# #         )
# #         self.async_model = AsyncOpenAI(
# #             base_url=os.environ["LOCAL_MODEL_BASE_URL"],
# #             api_key=os.environ["NVIDIA_API_KEY"],
# #             timeout=120,
# #             max_retries=5,
# #         )

# #     def generate(self, prompt: str) -> str:
# #         res = self.model.chat.completions.create(
# #             model=self.model_name,
# #             messages=[{"role": "user", "content": prompt}],
# #             temperature=self.temperature,
# #             max_tokens=512,
# #         )
# #         return res.choices[0].message.content

# #     async def a_generate(self, prompt: str) -> str:
# #         res = await self.async_model.chat.completions.create(  # ← async client
# #             model=self.model_name,
# #             messages=[{"role": "user", "content": prompt}],
# #             temperature=self.temperature,
# #             max_tokens=512,
# #         )
# #         return res.choices[0].message.content

# #     def get_model_name(self):
# #         return self.model_name   




# class DatasetPrep:
#     def __init__(self):
#         self.dataset_path = "Rakesh7n/ncrt_eval_outputs"

#     def combine_cols(self,example):

#         ans = example['Answer']
#         exp = example['Explanation']
#         return {"reference": f"Answer: {ans}\nExplanation: {exp}"}   

#     def load_test_data(self):            
#         ds = load_dataset(str(self.dataset_path), split="train") 
#         ds=ds.map(self.combine_cols)
#         ds=ds.select(range(1))
#         ds=ds.remove_columns(['Explanation','Question','Answer','messages'])
#         return ds

# from pydantic import BaseModel

# class ModelEvaluation:

#     def __init__(self, config: ModelEvaluationConfig):
#         self.config = config
#         self.llm = LocalModel(
#            LLMConfig("meta/muse-glimmer-30b",
#             0.0,)
#         )
#         # self.llm = NIMModel("meta/muse-glimmer-30b")
#         self.ds=DatasetPrep().load_test_data()

#     def metrics(self):
#         correctness_metric = GEval(
#         name="Correctness",
#         model=self.llm,
#         evaluation_steps=[
#             "Check whether the facts in 'actual output' contradict any facts in 'expected output'",
#             "Heavily penalize omission of key details from 'expected output'",
#             "Vague language or contradicting opinions are OK, but factual errors are not"
#         ],
#         evaluation_params=[ LLMTestCaseParams.ACTUAL_OUTPUT,  LLMTestCaseParams.EXPECTED_OUTPUT],
#         threshold=0.5,
#         )
#         relevancy_metric = AnswerRelevancyMetric(
#                 model=self.llm,
#                 threshold=0.5,
#             )
#         return correctness_metric, relevancy_metric
    
#     def build_test_cases(self,output_col):
#         test_cases = []
#         for i in range(len(self.ds)):
#             test_cases.append(LLMTestCase(
#                 input=self.ds["formatted_prompt"][i],
#                 actual_output=self.ds[output_col][i],
#                 expected_output=self.ds["reference"][i],
#             ))
#         return test_cases



#     def run_evaluation(self):
#         wandb.init(project="qwe3_4b_adapter-eval", name="adapter_evaluation_run")
#         correctness, relevancy = self.metrics()
#         test_cases = self.build_test_cases("adapter_output")
#         logger.info(f"Evaluating adapter on {len(test_cases)} samples...")
#         results = evaluate(test_cases=test_cases, metrics=[correctness, relevancy])

#         scores = {}
#         json_results = []

#         for i, tc_result in enumerate(results):
#             sample_entry = {
#                 "sample_id": i,
#                 "input": tc_result.test_case.input,
#                 "actual_output": tc_result.test_case.actual_output,
#                 "expected_output": tc_result.test_case.expected_output,
#                 "metrics": {},
#             }
#             for m in tc_result.metrics_data:
#                 scores.setdefault(m.name, []).append(m.score)
#                 sample_entry["metrics"][m.name] = m.score
#             json_results.append(sample_entry)

#         # Log averages to wandb (not per-sample in a loop)
#         for name, vals in scores.items():
#             avg = sum(vals) / len(vals)
#             pass_rate = sum(1 for v in vals if v >= 0.5) / len(vals)
#             logger.info(f"{name}: avg={avg:.3f}, pass_rate={pass_rate:.1%}")
#             wandb.log({f"{name}_avg": avg, f"{name}_pass_rate": pass_rate})

#         # Log per-sample table to wandb (one call, not 500)
#         import wandb as _wandb
#         table_data = [[e["sample_id"], e["input"][:100], e["metrics"].get("Correctness", 0), e["metrics"].get("Answer Relevancy", 0)] for e in json_results]
#         table = _wandb.Table(data=table_data, columns=["id", "input", "correctness", "relevancy"])
#         wandb.log({"sample_table": table})

#         output_path = os.path.join(self.config.root_dir, "evaluation_results.json")
#         with open(output_path, "w", encoding="utf-8") as f:
#             json.dump(json_results, f, indent=2, ensure_ascii=False)

#         logger.info(f"Results saved to {output_path}")
#         wandb.finish()   



#     # def run_evaluation(self):
#     #     wandb.init(
#     #         project="adapter-eval",
#     #         name="adapter_evaluation_run"
#     #     )
#     #     # self.load_test_data()
#     #     correctness, relevancy = self.metrics()
#     #     test_cases = self.build_test_cases("adapter_output")
#     #     logger.info(f"Evaluating adapter on {len(test_cases)} samples...")
#     #     results = evaluate(
#     #         test_cases=test_cases,
#     #         metrics=[correctness, relevancy],
#     #         # print_results=True,
#     #     )
#     #     # Extract scores
#     #     scores = {}
#     #     json_results = [] 
#     #     for tc_result in results:
#     #         sample_entry = {
#     #             "sample_id": tc_result.test_case_id,
#     #             "input": tc_result.test_case.input,
#     #             "actual_output": tc_result.test_case.actual_output,
#     #             "expected_output": tc_result.test_case.expected_output,
#     #             "metrics": {}
#     #         }
#     #         for m in tc_result.metrics_data:
#     #             scores.setdefault(m.name, []).append(m.score)
#     #             sample_entry["metrics"][m.name] = m.score
#     #             wandb.log({
#     #                 "sample_id": tc_result.test_case_id,
#     #                 f"{m.name}_score": m.score
#     #             })
#     #         json_results.append(sample_entry)
#     #     logger.info("=== Adapter Results ===")

#     #     for name, vals in scores.items():
#     #         avg = sum(vals) / len(vals)
#     #         pass_rate = sum(1 for v in vals if v >= 0.5) / len(vals)
#     #         logger.info(f"{name}: avg={avg:.3f}, pass_rate={pass_rate:.1%}")   
#     #         wandb.log({
#     #             f"{name}_avg": avg,
#     #             f"{name}_pass_rate": pass_rate
#     #         })

#     #     output_path = os.path.join(self.config.root_dir, "evaluation_results.json")
#     #     with open(output_path, "w", encoding="utf-8") as f:
#     #         json.dump(json_results, f, indent=2, ensure_ascii=False)

#     #     logger.info(f"Results saved to {output_path}")
#     #     wandb.finish()
# # --- Setup ---
# # os.environ["OPENAI_API_KEY"] = "your-api-key"  # or use `deepeval login`
# # base_cases = build_test_cases("base_output")
# # adapter_cases = build_test_cases("adapter_output")

# # import os
# # import torch
# # import wandb
# # from src.NcrtPhysicsQwen3.logger import logger
# # from src.NcrtPhysicsQwen3.config.configuration import ModelEvaluationConfig
# # from transformers import AutoTokenizer, AutoModelForCausalLM
# # from peft import PeftModel
# # from datasets import load_from_disk, Dataset
# # import tqdm
# # import json
# # import numpy as np



# # class ModelEvaluation:
# #     def __init__(self, config: ModelEvaluationConfig):
# #         self.config = config
# #         self.tokenizer = AutoTokenizer.from_pretrained(config.base_model_path)
# #         self.tokenizer.pad_token = self.tokenizer.eos_token


# # #     def generate_response(self,model,prompt,):
    
# # #     def evaluate(self):        
# # #         ds=load_from_disk(self.config.data_path)
# # #         base_model = AutoModelForCausalLM.from_pretrained(self.config.base_model_path)
# # #         # tokenizer= AutoTokenizer.from_pretrained(self.config.base_model_path)
# # #         adapter = PeftModel.from_pretrained(base_model, self.config.adapter_path)
# # #         return
```
